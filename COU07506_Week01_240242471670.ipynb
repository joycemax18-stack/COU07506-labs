Objective:to generate and plot elementary discrete-time signals.

```python
# Generate discrete-time index
n = np.arange(-5, 11)

# Unit impulse
delta = (n == 0).astype(float)

# Unit step
step = (n >= 0).astype(float)

# Unit ramp
ramp = n * step

# Real exponential
expo = 0.8**n * step

# Plot all four signals
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
fs, x = wavfile.read("my_voice.wav")

print("Sampling frequency:", fs, "Hz")
print("Duration:", len(x)/fs, "seconds")
print("Total samples:", len(x))
```

```python
# Convert audio to floating-point values
x = x.astype(float)

# Normalize audio
x = x / np.max(np.abs(x))

# Generate time axis
t = np.arange(len(x)) / fs

# Plot voice waveform
plt.figure(figsize=(12, 4))
plt.plot(t, x)

plt.title("My Voice Recording")
plt.xlabel("Time (seconds)")
plt.ylabel("Normalized Amplitude")
plt.grid()
plt.show()
```
